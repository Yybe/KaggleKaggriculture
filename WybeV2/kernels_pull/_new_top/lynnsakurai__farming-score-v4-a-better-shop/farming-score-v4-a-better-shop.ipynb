## Failure mechanism

For a near-mirror state, the original controller sets the adaptive budget to zero:

$$
B_i^{\mathrm{mirror}}=0.
$$

Because this budget is checked before urgency, it also disables adaptive selling when the Shed reaches its hard threshold or the episode enters its final tail. The urgency calculation therefore cannot affect a mirror-latched policy.

## Bounded correction

Define

$$
U_t=\mathbf{1}\!\left[S_t\ge 88\;\lor\;t\ge 704\right],
$$

where \(S_t\) is total Shed stock. The mirror budget becomes

$$
B_i^{\mathrm{mirror}}(t)=
\begin{cases}
0, & U_t=0,\\
6, & U_t=1.
\end{cases}
$$

The six-unit lifetime cap applies independently to `STRAWBERRY`, `MILK`, and `WOOL`. It is large enough to relieve pressure but remains substantially below the ordinary eighteen-unit budget.

## Preserved behavior

- Near-mirror behavior is unchanged before urgency.
- Non-mirror behavior is unchanged on every turn.
- The production trace, weed repair, reserves, price gates, flow estimator, tranche sizes, order ordering, and exception fallback are unchanged.
- The submission remains a single standalone `main.py`.

The source programme is derived from the supplied V29-R1 notebook. Its public-trace attribution and evaluation statements remain attributable to that source; this bounded correction requires fresh closed-loop evaluation.
